```python
for p in [1, 1.25, 1.5, 1.75, 2, 3, 10, 25, 500]:

    x_list = []
    y_list = []
    for t in np.linspace(0, 2*np.pi, 1000):

        x = np.cos(t)
        y = np.sin(t)

        r = (abs(x)**p + abs(y)**p)**(1/p)

        x_list.append(x/r)
        y_list.append(y/r)
        
        
    plt.figure(figsize=(3, 3))
    plt.plot(x_list, y_list, '-')
    plt.xticks([-1, -0.5, 0 , 0.5, 1])
    plt.yticks([-1, -0.5, 0 , 0.5, 1])
    plt.axis('equal')
    plt.grid()
    plt.title('p = ' + str(p))
    plt.show()
```
